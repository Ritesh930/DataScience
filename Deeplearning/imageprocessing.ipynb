```python
#pillow
from PIL import Image
img1=Image.open(r"C:\Users\yadav\OneDrive\Pictures\car.webp")
img1.show()
# img1.save("car image")
print(img.shape)  #shape of image   
print(img.size)  #size of image
print(img.dtype)  #image data type
```

```python
##scaling---changes the size of an image either enlarging or shrinking
# scaled= cv2.resize(img,(new_width,new_height))
scaled_image=cv2.resize(img,(200,200))
scaled_image=cv2.resize(img,(0,0),fx=0.5,fy=0.5)
plt.imshow(scaled_image)
```

```python
import numpy as np
rows,col=img.shape[:2]
tx,ty=1000 , 500 # transalation factor
M=np.float32([[1,0,tx],[0,1,ty]])
translated_image = cv2.warpAffine(img, M, (col, rows))
plt.imshow(translated_image)
```

```python
#rotation
center=(col/2,rows/2)  #center of image
angle=45
scale=0.5  #scale used for zoom in or zoom out
```

```python
#flipping---mirror image
flip_image=cv2.flip(img,0)  #flip image vertically
plt.imshow(flip_image)
```

```python
#cropping of image--slicing an array
crooped_image=img[100:200,200:300]
plt.imshow(crooped_image)
```

```python
#converting color image into black and white
img2=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.subplot(1,2,1)
plt.imshow(img1)
plt.title("original image")
plt.subplot(1,2,2)
plt.imshow(gray,cmap="gray")
plt.title("black and white image")
plt.show()
```

```python
#histograms---distribution of pixel intensity value
color=["b","g","r"]
for i,col in enumerate(color):
    hist=cv2.calcHist([img],[i],None,[256],[0,256])
    plt.plot(hist,color=col)
    plt.xlim([0,256])
plt.show()
```

```python
#erosion--remove white pixels from the boundary of image
#dilation---opposite of erosion
```

```python
kernel=np.ones((5,52),np.uint8)
erosion=cv2.erode(img,kernel,iterations=1)
dilate=cv2.dilate(img,kernel,iterations=1)
plt.imshow(erosion,cmap="gray")
```
