```python
import pandas as pd

df=pd.read_csv('Crop_recommendation.csv')  #reading the dataset
df.head()
```

```python
label=df.groupby('label')   #grouping by label
label.describe()
```

```python
label.get_group("rice") #this will give all the values of rice
```

```python
label[["humidity"]].mean() #this will give the mean of humidity
```

```python
labels = list(label.mean().index)
labels
```

```python
nitrogen = list(label[["N"]].mean().values.ravel())  #this will give the mean of nitrogen
nitrogen    
```

```python
import matplotlib.pyplot as plt

plt.bar(x=labels, height=nitrogen, ec="k")  #this will plot the bar graph
plt.xticks(rotation=90)    #this will rotate the labels
plt.show()
```

```python
x=df.drop("label",axis=1) #dropping the label column
```

```python
##standard scalar---normalise---outliers se deal karne k liye
from sklearn.preprocessing import StandardScaler    #this will normalise the data
sc=StandardScaler()    #this will normalise the data

x_transformed=sc.fit_transform(x)  #this will normalise the data

x_transformed.shape
```

```python
desc = pd.DataFrame(x_transformed, columns=x.columns).describe()  # this will create a dataframe with column names
desc.head()
```

```python
df.describe()  #this will describe the data
```

```python
desc_formatted=desc.applymap(lambda x: f"{int(x):d}")  #this will format the data
desc_formatted.head()
```

```python
from sklearn.cluster import KMeans  #
kmeans=KMeans()  #by default 8 hota hai
kmeans.fit(x_transformed)
```

```python
label = kmeans.labels_  #this will give the labels
```

```python
# Assign cluster labels to a new column in df
df["cluster_8"] = label
df["cluster_8"].value_counts()
```

```python
df[df["label"]=="rice"]["cluster_8"].value_counts()  #this will give the count of rice
```

```python
from sklearn.metrics import silhouette_score
##range ==  -1 to 1
silhouette_score(x_transformed,label)
```

```python
kmeans.inertia_  #this will give the inertia
```

```python
WCSS={}                       #this will create a dictionary
for i in range(5,51):            #this will create a loop
    kmeans=KMeans(n_clusters=i)     #this will create a KMeans object
    kmeans.fit(x_transformed)
    WCSS[i]=kmeans.inertia_
WCSS
```

```python
plt.figure(figsize=(20,8))    #this will create a figure
plt.plot(WCSS.keys(),WCSS.values(),marker="o")     #this will plot the graph
plt.xticks(list(WCSS.keys()))     #this will show the keys
plt.grid(True)  #this will show the grid
plt.show()
```
